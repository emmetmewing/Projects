## Methodology

To model the bubonic plague, we will first take the SEIR model, which will house our modeling of the various spreads. We will then model and graph the bubonic plague with the values associated with the modern research and see how that compares to historical trends. Then we will modify variables to see which single variable will lead to the greatest increase in the bubonic plague!

For this model, I will use a model to compartmentalize the population into Susceptible, Exposed, Infected, and Recovered/Dead groups. This model is known as the SEIR model. Unlike the SIR model, this takes into account the extra exposed group to better reflect the model

I will use the SEIR model, which is the Susceptible, Exposed, Infected, and Recovered/Dead. This model allows us to set various parameters for initial conditions, such as the number of susceptible people, which is the population minus the initial population. There is then the exposed, these are those who have the disease and are the start. There is also the number of infected and the recovered/dead, which are those who have had the disease. We also want the recovered and dead populations. Then there are parameters of beta, which is the transmission rate, sigma, which is the incubation rate, gamma, which is the removal rate, the mu, which is the background death rate, such as those who die without the disease, alpha, which is the disease death rate, and the waning immunity rate, which is the rate people lose immunity. 

Note - to account for seasonality, we will need to have the beta change over time. We can do this by having a sine function inside of it to deal with the seasonality for beta.

The equations we will use are:
- dSdt = mu * N - beta * I * S / N + omega * R - mu * S
- dEdt = beta * I * S / N - sigma * E - mu * E
- dIdt = sigma * E - gamma * I - (mu + alpha) * I
- dRdt = gamma * I - omega * R - mu * R

where E is the number of exposed, I is the number of expected, R is the number of recovered, and S is the number of susceptible.

When we are writing our code, we will have new copies of each variable so that we do not get any confusion when running it. Because we are modeling this with lots of different variables being changed, we want to ensure each variable is the same inside of its own context, so because of that, we will have many new variables.



```python
# importing the modules that I want
import numpy as np
from scipy.integrate import solve_ivp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# defining the model in terms of equations

# this is the base model that we will be using where we want to estimate the spread in Florence
def SEIR(time, current_values, mu, beta, omega, sigma, gamma, alpha):

    #what each parameter is
    # mu is the background death rate
    # beta is the transmission rate
    # omega is the waning immunity rate
    # sigma is the progression rate
    # gamma is the recovery rate
    # alpha is the mortality rate
    
    #unpacking the current values
    S, E, I, R = current_values
    
    #getting the population size
    N = S + E + I + R #this is because N is the total population and it is the sum of all of the values
    
    # now the derivatives
    dSdt = mu * N - beta * I * S / N + omega * R - mu * S
    dEdt = beta * I * S / N - sigma * E - mu * E
    dIdt = sigma * E - gamma * I - (mu + alpha) * I
    dRdt =  gamma * I - omega * R - mu * R
    
    return dSdt, dEdt, dIdt, dRdt
```

```python
# now we want to calculate various values for the Bubonic plague 

# from reviewing papers about the Bubonic plague, I have determined that
mu = 0.000039
omega = 1/365 # implying they will have a full year of immunity
sigma = 1/10 # (transmission rate is 10 days, generous so 1/10 is the progression rate)
gamma = 1/7 # this implies that the recovery rate is about 14%, which is close enough, but this might be a bit larger than it actually is
alpha = 45/100 # this is the mortality rate, sitting around 45% the mean of some WHO estiamtes

# with beta, we are plugging it into an equation to calculate the beta, because beta is weird
# to calculate beta, we will take R naught and put that in an equation
R_naught = 1.746
beta = R_naught * (sigma + mu) * (gamma + mu + alpha) / sigma 

# making the initial values, starting with 100000 total people, 10 infected, and 0 for recovered and 5 for exposed.
initial_values = [99985, 5, 10, 0]

# setting up the time array, where we will have a year
time_array = np.arange(0,721, 1)

initial_solutions = solve_ivp(SEIR, (0, 720), initial_values, t_eval=time_array, args=[mu, beta, omega, sigma, gamma, alpha])

S_values = initial_solutions.y[0,:]
E_values = initial_solutions.y[1,:]
I_values = initial_solutions.y[2,:]
R_values = initial_solutions.y[3,:]
```

```python
# now we want to plot it!
plt.plot(time_array, S_values, label="Susceptible")
plt.plot(time_array, E_values, label="Exposed")
plt.plot(time_array, I_values, label="Infected")
plt.plot(time_array, R_values, label="Recovered")

plt.title("Estimated Bubonic Plague Spread using Estimated Values")
plt.legend()
plt.xlabel("Days Since Start")
plt.ylabel("Population")
plt.show()
```

Now the next task to account for is to add seasonality to the model. To add seasonality, we will make use of the trigonometric functions. To do this, we will follow the tick season, so that means we need to assume that day 0 is January 1st, just to make coordinating the seasonality better.

To determine the argument for sine, I had to determine how I wanted it to spread. I want the spread to be less during the winter, and more during the summer, so I chose sine. I then wanted to shift sine a bit, so I moved it over by pi/2. The last factor was to determine the period of sine, and because we are doing it peaking in the summer, we needed it to osciallted every half year, giving us our equation for the seasonality.

```python
time_array = np.arange(0,721, 1)

seasonality = np.sin(2*np.pi/365*time_array-np.pi/2)

plt.plot(time_array, seasonality)
plt.title("A Demonstration of the Seasonality")
plt.xlabel("Day")
plt.ylabel("Seasonal Effect")
```

Now that I have made the equation to determine seasonality, I now need to find a way to make this effect the transmission rate, which will take making a new function. Additionally, beta will no longer be part of the parameters and it will be calculated during the function, as it is taking the seasonality into effect.

```python
# here we will modify the function to take into account beta fluctuating, so we will just be adding a line to keep defining beta


def SEIR_seasonal(time, current_values, mu, omega, sigma, gamma, alpha, R_naught, seasonality_multiplier):

    #what each parameter is
    # mu is the background death rate
    # beta is the transmission rate
    # omega is the waning immunity rate
    # sigma is the progression rate ( 1/ sigma would be the transmission rate time for infectious to become sick)
    # gamma is the recovery rate
    # alpha is the mortality rate

    #unpacking the current values
    S, E, I, R = current_values

    # we are doing the calculation of beta in here, and we have a seasonality multiplier to it to 
    #avoid having a result of 0 where the season has no effect because that does not really
    # make sense as the disease will still spread at the low points
    beta = ((R_naught * (sigma + mu) * (gamma + mu + alpha) / sigma) * (1 + seasonality_multiplier * np.sin(2*np.pi/365*time-np.pi/2)))

    
    #getting the population size
    N = S + E + I + R #this is because N is the total population and it is the sum of all of the values
    
    # now the derivatives
    dSdt = mu * N - beta * I * S / N + omega * R - mu * S
    dEdt = beta * I * S / N - sigma * E - mu * E
    dIdt = sigma * E - gamma * I - (mu + alpha) * I
    dRdt =  gamma * I - omega * R - mu * R
    
    return dSdt, dEdt, dIdt, dRdt
```

```python
# now we want to calculate various values for the Bubonic plague 

# from reviewing papers about the Bubonic plague, I have determined that
mu_t = 0.000039
omega_t = 1/365 # implying they will have a full year of immunity
sigma_t = 1/10 # (transmission rate is 10 days, generous so 1/10 is the progression rate)
gamma_t = 1/7
alpha_t = 45/100

# with beta, we are plugging it into an equation to calculate the beta, because beta is weird
# to calculate beta, we will take R naught and put that in an equation
R_naught_t = 1.746

# I am doing this to make it so that the seasonality is not too strong, but it is strong enough to pose an effect. It also makes it so that
# the seasons during the winter does not completely die
seasonality_t = 0.3

# making the initial values, starting with 100000 total people, 10 infected, and 0 for recovered and 5 for exposed.
initial_values_t = [99985, 5, 10, 0]

# setting up the time array, where we will have a year
time_array = np.arange(0,721, 1)

# here we are solving and unpacking the values
initial_solutions_t = solve_ivp(SEIR_seasonal, (0, 720), initial_values_t, t_eval=time_array, args=[mu_t, omega_t, sigma_t, gamma_t, alpha_t, R_naught_t, seasonality_t])

S_values_t = initial_solutions_t.y[0,:]
E_values_t = initial_solutions_t.y[1,:]
I_values_t = initial_solutions_t.y[2,:]
R_values_t = initial_solutions_t.y[3,:]

# now we want to plot it!
plt.plot(time_array, S_values_t, label="Susceptible")
plt.plot(time_array, E_values_t, label="Exposed")
plt.plot(time_array, I_values_t, label="Infected")
plt.plot(time_array, R_values_t, label="Recovered")

plt.title("Plauge Spread: Seasonality")
plt.legend()
plt.xlabel("Days Since Start")
plt.ylabel("Population")
plt.show()
```
