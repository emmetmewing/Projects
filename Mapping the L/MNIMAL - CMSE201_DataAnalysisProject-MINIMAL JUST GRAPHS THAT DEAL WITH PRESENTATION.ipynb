```python
def station_percent_growth(station):

    #getting the starting value
    mask_start = L_station_list['month_beginning'] == 2001
    mask2_start = L_station_list['stationame'] == station
    start = L_station_list[mask_start]
    start = start[mask2_start]
    start_value = start.iloc[0,2]
    
    #getting the ending value
    mask_end = L_station_list['month_beginning'] == 2019
    mask2_end = L_station_list['stationame'] == station
    end = L_station_list[mask_end]
    end = end[mask2_end]
    end_value = end.iloc[0,2]
   
    #now calculating the percent growth
    prct_growth = (end_value - start_value) / start_value * 100

    return prct_growth
    
    

def community_area_percent_growth(community_area):
    # this gets the population in 2000
    start_pop = community_areas_clean_dataframe[community_area]
    start_pop = start_pop.iloc[0]

    # this gets the end value in 2020
    end_pop = community_areas_clean_dataframe[community_area]
    end_pop = end_pop.iloc[-1]

    # this calculates the population growth percentage
    pop_growth = (end_pop - start_pop) / start_pop * 100    

    #this then returns the value
    return pop_growth
```
