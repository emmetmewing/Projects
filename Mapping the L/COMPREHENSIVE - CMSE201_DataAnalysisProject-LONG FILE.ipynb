Additionally, I will write a percent change function to help me calculate the percent changes that are discussed in the discussion section. I do not want to have the percent change for every single community area, as that would add way too much information to the brief discussions, so instead, this function will only be used for the few community areas and will be used in isolation.

Additionally, we only really care about certain values, so we will be calculating the percent growth from the station from the years 2001 to 2019, stopping before COVID, so we do not need that to show the change, as COVID was an irregular event.

For the population, we will take from 2000 to 2020, as we have data for those years, and we want to match percent growth to population growth closely. I chose 2020 as the later year because that was for a census, so it is a more exact value instead of a calculated one.

This also gives an error, the same error as the Correlation function, but that error comes because the L_station_list filters them together, so when we are filtering it, we are removing many indexes.

This function is used to calculate the values that are present in the discussion section. It is used by calling them with the station or community area.


```python
def station_percent_growth(station):

    #getting the starting value
    mask_start = L_station_list['month_beginning'] == 2001
    mask2_start = L_station_list['stationame'] == station
    start = L_station_list[mask_start]
    start = start[mask2_start]
    start_value = start.iloc[0,2]
    
    #getting the ending value
    mask_end = L_station_list['month_beginning'] == 2019
    mask2_end = L_station_list['stationame'] == station
    end = L_station_list[mask_end]
    end = end[mask2_end]
    end_value = end.iloc[0,2]
   
    #now calculating the percent growth
    prct_growth = (end_value - start_value) / start_value * 100

    return prct_growth
    
    

def community_area_percent_growth(community_area):
    # this gets the population in 2000
    start_pop = community_areas_clean_dataframe[community_area]
    start_pop = start_pop.iloc[0]

    # this gets the end value in 2020
    end_pop = community_areas_clean_dataframe[community_area]
    end_pop = end_pop.iloc[-1]

    # this calculates the population growth percentage
    pop_growth = (end_pop - start_pop) / start_pop * 100    

    #this then returns the value
    return pop_growth
```
